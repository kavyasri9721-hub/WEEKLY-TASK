```python
#number of rows before removing null values
print("rows before removing null values:", df.shape[0])

#remove rows containing null values
df = df.dropna()

#number of rows after removing null values
print("Rows after removing null values:",df.shape[0])
```

```python
print("number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
print("dataset shape removing duplicates", df.shape)
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df[["Order Date", "Ship Date"]].dtypes
```

```python
text_columns = ["Category", "Sub-Category", "Segment"]

for column in text_columns:
    df[column] = (
        df[column]
        .astype(str)
        .str.strip()
        .str.title()
    )
```

```python
for column in text_columns:
    print("\n", column)
    print(df[column].unique())
```

```python
print("Final Shape:", df.shape)

print("\nTotal Missing Values:", df.isnull().sum().sum())
print("Total Duplicate Rows:", df.duplicated().sum())
print("\nData Types:")
print(df.dtypes)
```

```python
numerical_features = [
    "Sales",
    "Quantity",
    "Discount",
    "Profit"
]

df[numerical_features].describe()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(
    data=df,
    x="Category",
    y="Sales",
    estimator="sum"
)
plt.title("Category-wise Total Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    y="Sales"
)
plt.title("Sales Distribution")
plt.ylabel("Sales")

plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    y="Profit"
)
plt.title("Profit Distribution")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="Category",
    y="Profit"
)
plt.title("Profit Distribution")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df,
    x="Discount",
    y="Profit"
)
plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

```python
df["Profit Margin"] = (df["Profit"]/df["Sales"])*100
df[["Sales", "Profit","Profit Margin"]].head()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit Margin"
)
plt.title("Discount vs Profit Margin")
plt.xlabel("Discount")
plt.ylabel("Profit Margin(%)")

plt.show()
```

```python
numerical_df = df.select_dtypes(include=np.number)

correlation_matrix = numerical_df.corr()

correlation_matrix
```

```python
plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Correlation Heatmap of Numerical Attributes")
plt.show()
```
