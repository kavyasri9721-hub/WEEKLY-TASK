```python
categorical_columns = [
    'gender',
    'race/ethnicity',
    'parental level of education',
    'lunch',
    'test preparation course'
]

for column in categorical_columns:
    print("\n", column)
    print(df[column].value_counts())
```

```python
score_columns = [
    'math score',
    'reading score',
    'writing score'
]

df[score_columns].describe()
```

```python
df['Total Score'] = (
    df['math score'] +
    df['reading score'] +
    df['writing score']
)
```

```python
df[['math score', 'reading score', 'writing score', 'Total Score']].head()
```

```python
df['Average Score'] = df[
    ['math score', 'reading score', 'writing score']
].mean(axis=1)
```

```python
df[['Total Score', 'Average Score']].head()
```

```python
df['Percentage Performance'] = (
    df['Total Score'] / 300
) * 100
```

```python
df[['Total Score', 'Average Score', 'Percentage Performance']].head()
```

```python
df['Percentage Performance'] = (
    df['Total Score'] / 300
) * 100
```

```python
df[['Total Score', 'Average Score', 'Percentage Performance']].head()
```

```python
df['Pass/Fail'] = np.where(
    (df['math score'] >= 40) &
    (df['reading score'] >= 40) &
    (df['writing score'] >= 40),
    'Pass',
    'Fail'
)
```

```python
df['Pass/Fail'].value_counts()
```

```python
print("Mean:")
print(df[score_columns].mean())

print("\nMedian:")
print(df[score_columns].median())

print("\nStandard Deviation:")
print(df[score_columns].std())
```

```python
print(df[score_columns].quantile([0.25, 0.50, 0.75]))
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df[score_columns]
)

plt.title("Score Distribution and Outliers")
plt.xlabel("Subjects")
plt.ylabel("Score")

plt.show()
```

```python
parental_avg = df.groupby(
    'parental level of education'
)[score_columns].mean()

parental_avg
```

```python
parental_avg.plot(
    kind='bar',
    figsize=(12, 6)
)

plt.title("Average Scores by Parental Level of Education")
plt.xlabel("Parental Level of Education")
plt.ylabel("Average Score")
plt.xticks(rotation=45)
plt.legend(title="Subject")

plt.show()
```

```python
lunch_avg = df.groupby('lunch')[score_columns].mean()

lunch_avg
```

```python
lunch_avg.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title("Average Scores by Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subject")

plt.show()
```

```python
test_prep_avg = df.groupby(
    'test preparation course'
)[score_columns].mean()

test_prep_avg
```
