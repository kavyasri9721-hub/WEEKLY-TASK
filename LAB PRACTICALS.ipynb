PRACTICAL 2  usingconditional statements and loops

```python
# 1. Check Positive Number (if)
num = int(input("Enter a number: "))
if num > 0:
    print("Positive Number")

# 2. Even or Odd (if-else)
num = int(input("Enter another number: "))
if num % 2 == 0:
    print("Even Number")
else:
    print("Odd Number")

# 3. Grade (if-elif-else)
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 50:
    print("Grade: C")
else:
    print("Fail")

# 4. Login (Nested if-else)
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Incorrect Password")
else:
    print("Invalid Username")

# 5. Voting Eligibility (One-line if-else)
age = int(input("Enter your age: "))
print("Eligible to Vote" if age >= 18 else "Not Eligible to Vote")

# 6. For Loop
print("Numbers using for loop:")
for i in range(1, 6):
    print(i)

# 7. While Loop
print("Numbers using while loop:")
i = 1
while i <= 5:
    print(i)
    i += 1

# 8. Break Statement
print("Break Example:")
for i in range(1, 11):
    if i == 5:
        break
    print(i)

# 9. Continue Statement
print("Continue Example:")
for i in range(1, 11):
    if i == 5:
        continue
    print(i)

# 10. Pass Statement
print("Pass Example:")
for i in range(1, 6):
    if i == 3:
        pass
    print(i)

# 11. Match-Case Statement
day = int(input("Enter day number (1-7): "))

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Day")

print("Program Ended")
```

PRACTICAL 3 USER DEFINED FUNCTIONS

```python
# 1. No arguments, No return value
def greet():
    print("Hello!")

greet()

# 2. Arguments, No return value
def add(a, b):
    print("Sum =", a + b)

add(10, 20)

# 3. No arguments, Return value
def message():
    return "Welcome to Python"

print(message())

# 4. Arguments, Return value
def multiply(a, b):
    return a * b

print("Product =", multiply(5, 4))

# 5. Default argument
def student(name="Dharshini"):
    print("Name:", name)

student()
student("Ravi")

# 6. Keyword arguments
def details(name, age):
    print("Name:", name)
    print("Age:", age)

details(age=18, name="Anu")

# 7. Variable length arguments
def total(*marks):
    print("Total Marks =", sum(marks))

total(80, 90, 85)

# 8. Lambda function
square = lambda x: x * x
print("Square =", square(5))

# 9. Recursive function
def fact(n):
    if n == 1:
        return 1
    return n * fact(n - 1)

print("Factorial =", fact(5))

# 10. Nested function
def outer():
    def inner():
        print("Inner Function")
    inner()

outer()
```

PRACTICAL 4 CREATE A LIST 

```python
nums = [1, 2, 3, 4, 5]

print("Original List:", nums)
print("First Element:", nums[0])
print("Last Element:", nums[-1])

# Append
nums.append(6)
print("After Append:", nums)

# Insert
nums.insert(2, 10)
print("After Insert:", nums)

# Remove
nums.remove(10)
print("After Remove:", nums)

# Length
print("Length:", len(nums))

# Traverse
print("List Elements:")
for i in nums:
    print(i)

# List Comprehension
square = [x*x for x in nums]
even = [x for x in nums if x % 2 == 0]
odd = [x for x in nums if x % 2 != 0]
double = [x*2 for x in nums]

matrix = [[1, 2], [3, 4]]
flat = [x for row in matrix for x in row]

print("Squares:", square)
print("Even:", even)
print("Odd:", odd)
print("Double:", double)
print("Flattened List:", flat)
```

PRACTICAL 5 TUPLE

```python
t = (10, 20, 30, 40, 20)

print("Tuple:", t)
print("First Element:", t[0])
print("Last Element:", t[-1])
print("Length:", len(t))
print("Slicing:", t[1:4])
print("Count of 20:", t.count(20))
print("Index of 30:", t.index(30))

print("Tuple Elements:")
for i in t:
    print(i)

t2 = (50, 60)
print("Concatenated Tuple:", t + t2)

# Set
s = {1, 2, 3, 4}

print("Original Set:", s)

s.add(5)
print("After Add:", s)

s.remove(2)
print("After Remove:", s)

print("Length:", len(s))

print("Set Elements:")
for i in s:
    print(i)

s2 = {3, 4, 5, 6}

print("Union:", s | s2)
print("Intersection:", s & s2)
print("Difference:", s - s2)
print("Symmetric Difference:", s ^ s2)
```

PRACTICAL 6 DICTIONARY OPERATIONS

```python
student = {
    "Name": "Sirushtika",
    "Age": 18,
    "Course": "BSc DS & AI"
}

print("Original Dictionary:", student)

# Access value
print("Name:", student["Name"])

# Add new key
student["College"] = "SDNB Vaishnav College"

# Update value
student["Age"] = 19

# Remove key
student.pop("Course")

# Length
print("Length:", len(student))

# Keys
print("Keys:", student.keys())

# Values
print("Values:", student.values())

# Items
print("Items:", student.items())

# Traverse dictionary
print("Dictionary:")
for key, value in student.items():
    print(key, ":", value)

# Check key
if "Age" in student:
    print("Age key exists")

# Copy dictionary
copy_student = student.copy()
print("Copied Dictionary:", copy_student)

# Clear another dictionary
d = {"A": 1, "B": 2}
d.clear()
print("Cleared Dictionary:", d)

print("Updated Dictionary:", student)
```
