```python
df[
    (df["Age"]>60)&
    (df["Admission Type"] =="Emergency")
    ]
```

```python
# 1. Female patients with Diabetes
df[(df["Gender"] == "Female") & (df["Medical Condition"] == "Diabetes")]

# 2. Age < 30 and Elective admission
df[(df["Age"] < 30) & (df["Admission Type"] == "Elective")]

# 3. High Billing Amount > 40000 and Male
df[(df["Billing Amount"] > 40000) & (df["Gender"] == "Male")]

# 4. Cancer patients with A+ Blood Type
df[(df["Medical Condition"] == "Cancer") & (df["Blood Type"] == "A+")]

# 5. 3 conditions - Age > 50, Female, Insurance = Medicare
df[
    (df["Age"] > 50) & 
    (df["Gender"] == "Female") & 
    (df["Insurance Provider"] == "Medicare")
]

# 6. Using OR condition | - Age < 18 OR Age > 75
df[(df["Age"] < 18) | (df["Age"] > 75)]

# 7. Using NOT condition ~ - NOT Male (means Female)
df[~(df["Gender"] == "Male")]

# 8. Billing between 20000 to 30000
df[(df["Billing Amount"] >= 20000) & (df["Billing Amount"] <= 30000)]
```

```python
# 17. Create a New Feature - Length of Stay

df["Length of Stay"] = (
    pd.to_datetime(df["Discharge Date"]) - 
    pd.to_datetime(df["Date of Admission"])
).dt.days

df[["Date of Admission", "Discharge Date", "Length of Stay"]].head()
```

```python
# 18. Create a New Feature - Admission Year

df["Admission Year"] = pd.to_datetime(
    df["Date of Admission"]
).dt.year

df[["Date of Admission", "Admission Year"]].head()
```

```python
# 20. Average Billing Amount by Admission Year

df.groupby("Admission Year")["Billing Amount"].mean()
```
